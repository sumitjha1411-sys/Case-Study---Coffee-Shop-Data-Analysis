### 2: Recipe and Ingredient Usage Analysis

```python
recipe_usage_data.info()
```

```python
# Grouping total quantity of ingredients used for each item

quantity_data = recipe_usage_data.groupby('ing_id')['quantity'].sum()

# Calculate the total quantity

total_quantity = quantity_data.sum()

# Calculate the percentage for each ingredient

percentage_data = (quantity_data / total_quantity) * 100

#sorting data

percentage_data = percentage_data.sort_values(ascending=False)

percentage_data
```

### 3: Ingredient Cost Analysis

```python
# Importing recipe_ingredients table

ing_usage_data = pd.read_csv('/Users/sumit/Documents/Saga Case Study/recipe_ingredients.csv')

ing_usage_data
```

```python
# Calculating ingredient cost per item

ing_usage_data['total_cost'] = ing_usage_data['quantity'] * ing_usage_data['ing_price']

ing_usage_data
```

```python
# Aggregate ingredient cost per  item

item_cost = ing_usage_data.groupby('ing_id')['total_cost'].sum()

#sorting data

item_cost = item_cost.sort_values(ascending=False)

item_cost
```

```python
# Plot the total ingredient cost per Ingredient

item_cost.plot(kind='bar', title='Total Ingredient Cost by Ingredient quantity and price')
plt.show()
```

### 5: Staff Scheduling Analysis

```python
# Importing staff_rota_shift table


staff_data = pd.read_csv('/Users/sumit/Documents/Saga Case Study/staff_rota_shift.csv')

staff_data
```

```python
staff_data.info()
```

```python
# Calculating total hours worked per staff member

staff_data['start_time'] = pd.to_datetime(staff_data['start_time'])
staff_data['end_time'] = pd.to_datetime(staff_data['end_time'])

staff_data['shift_duration'] = (staff_data['end_time'] - staff_data['start_time']).dt.total_seconds() / 3600

staff_data
```

```python
# Grouping total hours worked by each staff 

staff_hours = staff_data.groupby('staff_full_name')['shift_duration'].sum()

staff_hours

# Plot total hours worked by each staff

staff_hours.plot(kind='bar', title='Total Hours Worked by Staff')
plt.show()
```

```python
# Count shifts per staff by day

staff_day_count = staff_data.groupby('staff_full_name')['day_of_week'].count()

staff_day_count
```

```python
staff_coverage = staff_data.groupby(['day_of_week', 'staff_full_name'])['shift_id'].count().unstack()

staff_coverage
```

### 6: Item Profitability Analysis

```python
# Importing cost_analysis table


profit_data = pd.read_csv('/Users/sumit/Documents/Saga Case Study/cost_analysis.csv')

profit_data
```

```python
# Calculating selling cost for each recipe

item_profit = profit_data.groupby('recipe_id')['item_selling_price'].sum()

item_profit
```

```python
item_profit.count()
```

```python
# Importing cost table


cost_data = pd.read_csv('/Users/sumit/Documents/Saga Case Study/cost.csv')

cost_data
```

```python
# Joining cost_data and item_profit tables

join = cost_data.merge(item_profit, on='recipe_id', how='inner')

join
```

```python
join.count()
```

```python
# calculating profit margin for each recipe

join['profit_margin'] = join['item_selling_price'] - join['total_receipe_cost']

join=join.sort_values(by='profit_margin', ascending=False)

join

```

```python
# Plotting Item cost vs selling price

plt.figure(figsize=(10, 6))
plt.scatter(join['total_receipe_cost'], join['item_selling_price'], c=join['profit_margin'])
plt.colorbar(label='Profit Margin')
plt.xlabel('Total Recipe Cost')
plt.ylabel('Item Selling Price')
plt.title('Total Recipe Cost vs Item Selling Price')
plt.grid(True)
plt.show()
```
